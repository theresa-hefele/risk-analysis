Before we dive into lists, let’s understand what an **iterable** is.  

In Python, an **iterable** is **any object that you can loop through** (or *iterate* over).  
This means you can go through its elements **one by one** using a `for` loop.  

✅ **Common iterables include:**
- **Strings** → `'hello'` (you can loop through each character)
- **Lists** → `[1, 2, 3]` (you can loop through each item)
- **Tuples**, **sets**, and **dictionaries** are also iterables.
