````python
# We need the follwing imports
import matplotlib.pyplot as plt
import pandas as pd
weather_fire_data = pd.read_csv("weather_fires.csv")
````

````python
# a) explore the data
print(weather_fire_data.head())
````

````python
# b) plot time series diagrams 

# Ensure Date is datetime
weather_fire_data["Date"] = pd.to_datetime(weather_fire_data["Date"], format="%Y%m%d")

# Filter for 2006
df_2006 = weather_fire_data[weather_fire_data["Date"].dt.year == 2006]

temp_cols = ['Akrotiri_Temp', 'Larnaka_Temp', 'Paphos_Temp']
humidity_cols = ['Akrotiri_RH', 'Larnaka_RH', 'Paphos_RH']

# ---- Plot Wildfires ----
plt.figure(figsize=(12, 4))
plt.scatter(df_2006["Date"], df_2006["Number"], label="Wildfires")
plt.title("Wildfires in 2006 (Cyprus)")
plt.xlabel("Date")
plt.ylabel("Number of Fires")
plt.legend()
plt.grid(True)
plt.show()

# ---- Plot Temperature ----
plt.figure(figsize=(12, 4))
for col in temp_cols:
    plt.plot(df_2006["Date"], df_2006[col], label=col)
plt.title("Daily Temperature in 2006")
plt.xlabel("Date")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.grid(True)
plt.show()

# ---- Plot Humidity ----
plt.figure(figsize=(12, 4))
for col in humidity_cols:
    plt.plot(df_2006["Date"], df_2006[col], label=col)
plt.title("Daily Relative Humidity in 2006")
plt.xlabel("Date")
plt.ylabel("Humidity (%)")
plt.legend()
plt.grid(True)
plt.show()
````

### c) Plot Normalized Frequency Diagrams

````python
# Let's look at the temperature distribution first

# We'll create a new column to use the average daily temperature across the 3 stations
weather_fire_data["Mean_Temp"] = weather_fire_data[temp_cols].mean(axis=1)

plt.figure(figsize=(10,6))
plt.hist(weather_fire_data["Mean_Temp"], bins=20, density=True) # density=True normalizes the histogram
plt.title("Normalized Temperature Distribution: All Days (2006–2007)")
plt.xlabel("Temperature (°C)")
plt.ylabel("Normalized Frequency")
plt.grid(True)
````

````python
# Create a flag for fire days
weather_fire_data["FireDay"] = weather_fire_data["Number"] > 0

# Separate fire vs no-fire days
temps_fire = weather_fire_data[weather_fire_data["FireDay"] == True]["Mean_Temp"]
temps_no_fire = weather_fire_data[weather_fire_data["FireDay"] == False]["Mean_Temp"]

plt.figure(figsize=(8,5))
plt.hist(temps_no_fire, bins=20, color='blue', density=True) # density=True normalizes the histogram
plt.title("Normalized Temperature Distributions: No Fire Days (2006–2007)")
plt.xlabel("Temperature (°C)")
plt.ylabel("Normalized Frequency")
plt.grid(True)

plt.figure(figsize=(8,5))
plt.hist(temps_fire, bins=20, color='red', density=True) 
plt.title("Normalized Temperature Distributions: Fire Days (2006–2007)")
plt.xlabel("Temperature (°C)")
plt.ylabel("Normalized Frequency")
plt.grid(True)
plt.show()
````

````python
# If we want to relate the fire days to the all days temperature distribution, we can stack the two distributions on top of each other:
plt.figure(figsize=(8,10))
plt.subplot(2,1,1)
plt.hist([temps_no_fire, temps_fire], density=True, stacked=True, bins=20, label=['No Fire Days', 'Fire Days'], color=['blue', 'red'])
plt.ylabel('Frequency')
plt.title('Normalized Frequency of All Days')
plt.legend()
plt.grid(True)
plt.subplot(2,1,2)
plt.hist([temps_no_fire, temps_fire], density=True, stacked=True, bins=20, label=['No Fire Days', 'Fire Days'], color=['blue', 'red'], cumulative=True)
plt.xlabel('Temperature (°C)')
plt.title('Cumulative Frequency of All Days')
plt.ylabel('Cumulative Frequency')
plt.grid(True)
plt.show()
````

### d) Determine the correlation between relative humidity and termperature and visualize it using a scatter plot

````python
# Identify temperature and humidity columns
temp_cols = ['Akrotiri_Temp', 'Larnaka_Temp', 'Paphos_Temp']
humidity_cols = ['Akrotiri_RH', 'Larnaka_RH', 'Paphos_RH']
stations = ['Akrotiri', 'Larnaka', 'Paphos']


mean_temperature = weather_fire_data[temp_cols].mean(axis=1)
mean_humidity = weather_fire_data[humidity_cols].mean(axis=1)
# Calculate correlation between mean temperature and mean humidity
corr_mean_temp_humidity = mean_temperature.corr(mean_humidity)

plt.scatter(mean_humidity, mean_temperature)
plt.ylabel('mean_temperature')
plt.xlabel('mean_humidity')
plt.title(f'Scatter plot for mean humidity and temperature, Correlation: {corr_mean_temp_humidity:.2f}')

for i, station in enumerate(stations):

    humidity = weather_fire_data[humidity_cols[i]]
    temperature = weather_fire_data[temp_cols[i]]

    plt.figure()
    plt.scatter(humidity, temperature)
    plt.ylabel('temperature')
    plt.xlabel('humidity')
    plt.title(f'Scatter plot for {station}, correlation: {temperature.corr(humidity):.2f}')
````
