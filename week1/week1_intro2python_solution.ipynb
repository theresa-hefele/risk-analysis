1. Create a dictionary that contains at least **three key-value pairs**.  
   - The **key** could be a category (like a name, subject, or item).  
   - The **value** could be a number or string related to it.  
2. Print the **entire dictionary** using an **f-string**.

💡 *Hints:*

- Looping with `.items()` gives you both key and value.


````python
# Define your dictionary
student_info = {
    "name": "Alex",
    "age": 21,
    "major": "Civil Engineering",
    "courses": ["Risk Analysis", "Math", "Architecture"]
}

# Print the full dictionary
print(f"Student information: {student_info}")

# Access and print a specific value
print(f"The student's name is: {student_info['name']}")

# Loop through the dictionary and print each key-value pair
print("Details:")
for key, value in student_info.items():
    print(f"{key}: {value}")
````

````python
# Data (times in minutes)
walking_time_to_station = 12   # time to walk to U-Bahn
bus_arrival_time = 5           # time until the next bus arrives
bus_travel_time = 6            # time bus takes to get to U-Bahn

# Decision using logical operators
total_bus_time = bus_arrival_time + bus_travel_time

if walking_time_to_station < total_bus_time:
    print("Better walk to the U-Bahn station — you'll arrive faster!")
elif walking_time_to_station == total_bus_time:
    print("Either walking or taking the bus will take the same time.")
else:
    print("Wait for the bus — it will get you there faster.")

# Extra check using AND / OR.
# This prints nothing: walking takes 12 minutes (not more than 15)
# and the bus arrives in 5 minutes (not more than 10), so neither condition is true.
if walking_time_to_station <= 15 and bus_arrival_time > 10:
    print("Even though the bus is coming, walking is still reasonable.")
elif walking_time_to_station > 15 or bus_arrival_time > 10:
    print("Walking might be tiring or the bus is too far away, plan accordingly.")
````

````python
import pandas as pd

# Original multi-row DataFrame
scores_dictionary = {
    "Alice": [85, 90, 88],
    "Ben": [92, 87, 94],
    "Chloe": [78, 82, 85]
}

scores_df = pd.DataFrame(scores_dictionary, index=["Test 1", "Test 2", "Test 3"])

print("Original DataFrame:")
print(scores_df)

# Access one column and one row
print("\nAlice's scores:")
print(scores_df["Alice"])
print("\nFirst test:")
print(scores_df.loc["Test 1"])

# Add a new test (row)
scores_df.loc["Test 4"] = [91, 89, 87]
print("\nUpdated DataFrame with Test 4 added:\n")
print(scores_df)

# Add a new student
scores_df["Walter"] = [95, 98, 97, 96]
print("\nWalter has joined the class!\n")
print(scores_df)

# Calculate average score per student (column-wise)
average_scores = scores_df.mean(axis=0)
print("\nAverage score per student:\n")

for student, avg in average_scores.items():
    print(f"{student}: {avg}")
````
